```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'oreder':[1,2,3,5,2],
    'product':['cup','plate','cup','plate','cup'],
    'price':[10,20,30,40,50],
    'quantity':[1,2,3,4,5],
    'customer':['ahmed','ali','ahmed','ali','ali']

}
df=pd.DataFrame(data)
df
```

```python
df['totall']=df['price']*df['quantity']
df
```

```python
df[['oreder','product','price']]
df
```

```python
df[df['product']=='cup']
```

```python
df[df['price']>20]
```

```python
df[df['totall']>50]
```

```python
df.groupby('price')['totall'].sum().sort_values(ascending=False)
```

```python
df.groupby('product')['totall'].sum()
```

```python
df.groupby('oreder')['totall'].sum().sort_values(ascending=False)
```

```python
df.groupby('product')['totall'].sum().sort_values(ascending=False)
```

```python
report=df.groupby('customer')['totall'].sum().sort_values(ascending=False)
report
```

```python
report=df.groupby('customer').agg({
    'totall':'sum',
    'quantity':'sum'
})
report
```

```python
sns.barplot(
    x=report.index,
    y=report['totall']
)
plt.title('total sales per customer')
plt.show()
```
