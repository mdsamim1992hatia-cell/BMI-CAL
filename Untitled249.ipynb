```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,101),
    'product':np.random.choice(['Apple','Ball',None,None],100),
    'quantity':np.random.randint(1,5,100),
    'customer':np.random.choice(['Alex','rolla',None,None,'ALEVV'],100),
    'price':np.random.randint(100,1000,100)
}
df=pd.DataFrame(data)
print("====raw data====")
print(df)
```

```python
df['product']=df['product'].fillna('None')
df
```

```python
df['customer']=df['customer'].fillna('None')
df
```

```python
df['total_sales']=df['quantity']*df['price']
df
```

```python
top_sales=df.sort_values(by='total_sales',ascending=False)
top_sales
```

```python
top_customer=df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
top_customer
```

```python
top_product=df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
top_product
```

```python
avg_sales=df.groupby('customer')['total_sales'].mean()
avg_sales
```

```python
avg_order=df.groupby('order')['total_sales'].mean()
avg_order
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_customer.index,y=top_customer.values)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Top Customers by Sales')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_product.index,y=top_product.values)
plt.xlabel('Product')
plt
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x=avg_sales.index,y=avg_sales.values)
plt.xlabel('Customer')
plt.ylabel('Average Sales')
plt.title('Average Sales per Customer')
```
