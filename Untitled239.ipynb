```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,51),
    'customer':np.random.choice(['alex',None,'ANAEC','asses'],50),
    'sales':np.random.randint(100,500,50),
    'product':np.random.choice(['laptop','mobile',None],50),
    'date':pd.date_range('2023-01-01',periods=50,freq='D'),
    'quantity':np.random.randint(1,10,50)
}
df=pd.DataFrame(data)
print("====RAW DATA====\n",df)
```

```python
df['product']=df['product'].fillna('unknown')
df['customer']=df['customer'].fillna('unknown')
print("===clean data===\n",df)


```

```python
df['total_sales']=df['sales']*df['quantity']
print("===total sales===\n",df)
```

```python
top_customer=df.groupby('customer')['total_sales'].sum().sort_values(ascending=False).head(5)
print("===top customer===\n",top_customer)
```

```python
daily_sales=df.groupby(df['date'].dt.date)['total_sales'].sum()
print("===daily sales===\n",daily_sales)
```

```python
top_order=df.nlargest(5,'total_sales')
print("===top order===\n",top_order)
```

```python
total_revenue=df['total_sales'].sum()
print("===total revenue===\n",total_revenue)
```

```python
customer_sales=df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
print("===customer sales===\n",customer_sales)
```

```python
product_sales=df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
print("===product sales===\n",product_sales)
```

```python
np.average_sales=df['total_sales'].mean()
print("===average sales===\n",np.average_sales)
```

```python
customer_avg=df.groupby('customer')['total_sales'].mean().sort_values
print("===customer avg===\n",customer_avg)
#
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='total_sales',data=df)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Total Sales by Customer')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='date',y='total_sales',data=df)
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.title('Total Sales Over Time')
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='product',y='total_sales',data=df)
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Total Sales by Product')
plt
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='total_sales',hue='product',data=df)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Total Sales by Customer and Product')
plt
```
