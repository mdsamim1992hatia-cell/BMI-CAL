```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1=pd.DataFrame({
    'key':['b','b','a','c','a','a','b'],
    'x1':range(7),
    'y1':range(7,0,-1)

})
print(df1)
```

```python
df2=pd.DataFrame({
    'key':['a','b','d'],
    'x2':range(3),
    'y2':range(3,0,-1)

})
print(df2)
```

```python
pd.merge(df1,df2,on='key',how='inner')
```

```python
pd.merge(df1,df2,on='key',how='inner')[['key','x1','y1']]
```

```python
pd.merge(df1,df2,on='key',how='outer')
```

```python
pd.merge(df1,df2,on='key',how='outer')[['key','x1','y1']]
```

```python
pd.merge(df1,df2,on='key',how='left')
```

```python
pd.merge(df1,df2,on='key',how='left')[['key','x1','y1']]
```

```python
pd.merge(df1,df2,on='key',how='right')
```

```python
pd.merge(df1,df2,on='key',how='right')[['key','x1','y1']]
```
