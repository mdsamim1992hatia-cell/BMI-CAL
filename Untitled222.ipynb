```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,11),
    'product':['A','B',None,'A','B','C',None,'B',None,'A'],
    'customer':['X','Y','Z',None,'Y',None,'X','Y',None,'X'],
    'quantity':[5,3,2,4,6,2,3,1,4,5],
    'price':[10,15,20,12,18,None,1855,555,None,8585],

}
df=pd.DataFrame(data)
print("===data frame===")
print(df)
```

```python
df.isna().sum()
```

```python
df.isna().sum().sum()
```

```python
df
```

```python
df['customer']=df['customer'].fillna('Unknown')
df['product']=df['product'].fillna('Unknown')
df['price']=df['price'].fillna(df['price'].mean())
print("===clean data===")
print(df)
```

```python
df['total_price']=df['quantity']*df['price']
print("===total price===")
print(df)
```

```python
product_sales=df.groupby('product')['total_price'].sum().sort_values(ascending=False)
print("===product sales===")
print(product_sales)
```

```python
customer_sales=df.groupby('customer')['total_price'].sum().sort_values(ascending=False)
print("===customer sales===")
print(customer_sales)
```

```python
order_price=df.groupby('order')['total_price'].sum().sort_values(ascending=False)
print("===order price===")
print(order_price)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=product_sales.index,y=product_sales.values)
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Total Sales by Product')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=customer_sales.index,y=customer_sales.values)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Total Sales by Customer')
plt.xticks(rotation=90)
plt.show()
```
