```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':[1,2,3,4,5],
    'product':['A','B','C',   None,'E'],
    'quantity':[10,15,8,12,20],
    'price':[100,50,75,60,90],
    'customer':['Alice','Bob','Charlie','David',None]
}
df=pd.DataFrame(data)
df
```

```python
df.isna().sum()
```

```python
df.isna().sum()/len(df)*100
```

```python
df.isna().sum().sum()
```

```python
df.shape
```

```python
df.notna().sum()
```

```python
df.notna().sum().sum()
```

```python
df['total_price']=df['quantity']*df['price']
df
```

```python
df['customer']=df['customer'].fillna('Unknown')
df
```

```python
df['product']=df['product'].fillna('Unknown')
df
```

```python
df.isna().sum()
```

```python
df.groupby('customer')['total_price'].sum()
```

```python
df.groupby('product').agg({'quantity':'sum','total_price':'sum'})
```

```python
df.groupby('customer').agg({'quantity':'sum','total_price':'sum'})
```

```python
df.groupby(['customer','product']).agg({'quantity':'sum','total_price':'sum'})
```

```python
df.groupby(['customer','product']).agg({'quantity':'sum','total_price':'sum'}).sort_values('total_price',ascending=False)
```

```python
df.groupby('customer').agg({'total_price':'sum'}).sort_values('total_price',ascending=False)
```

```python
top_customers=df.groupby('customer').agg({'total_price':'sum'}).sort_values('total_price',ascending=False).head(50)
top_customers
```

```python
top_products=df.groupby('product').agg({'quantity':'sum','total_price':'sum'}).sort_values('total_price',)
top_products
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_customers.index,y=top_customers['total_price'])
plt.xticks(rotation=90)
plt.xlabel('Customer')
plt.ylabel('Total Price')
plt.title('Top 50 Customers by Total Price')
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_products.index,y=top_products['total_price'])
plt.xticks(rotation=90)
plt.xlabel('Product')
plt.ylabel('Total Price')
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_products.index,y=top_products['quantity'])
plt.xticks(rotation=90)
plt.xlabel('Product')
plt.ylabel('Quantity')
```
