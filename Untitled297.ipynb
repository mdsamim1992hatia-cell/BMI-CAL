```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1=pd.DataFrame({
    'key':['a','b','c','d','e'],
    'a1':range(5)	,
    'b1':range(5,10)
})
print(df1)
```

```python
df2=pd.DataFrame({
    'key':['a','b','c'],
    'a2':range(3),
    'b2':range(3,6)
})
print(df2)
```

```python
pd.merge(df1,df2,on='key',how='inner')[['key','a1','a2']]
```

```python
df4=pd.merge(df1,df2,on='key',how='left')
print(df4)
```

```python
df4=pd.merge(df1,df2,on='key',how='left')[['key','a1','a2']]
print(df4)
```

```python
df5=pd.merge(df1,df2,on='key',how='right')
print(df5)
```

```python
df5=pd.merge(df1,df2,on='key',how='right')[['key','a1','a2']]
print(df5)
```

```python
df6=pd.merge(df1,df2,on='key',how='outer')
print(df6)
```

```python
df6=pd.merge(df1,df2,on='key',how='outer')[['key','a1','a2']]
print(df6)
```
