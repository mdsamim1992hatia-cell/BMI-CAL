```python
from itertools import product
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)
roads=[f'road{i}' for i  in range (1,7)]
product=[f'product{i}' for i  in range (1,26)]
```

```python
data={
    'road':np.random.choice(roads,size=500),
    'product':np.random.choice(product,size=500),
    'quantity':np.random.randint(1,10,size=500)
}
df=pd.DataFrame(data)
df
#
```

```python
print(df.head())
```

```python
road__product_sales=df.groupby(['road','product'])['quantity'].sum().reset_index()
road__product_sales
```

```python
top_product_per_road=road__product_sales.loc[road__product_sales.groupby('road')['quantity'].idxmax()]
top_product_per_road
```

```python
pivot_df=df.pivot_table(index='product',columns='road',values='quantity')
pivot_df
```

```python
plt.figure(figsize=(12,10))
sns.heatmap(pivot_df,cmap='YlGnBu',annot=True,fmt='.0f',cbar=False)
plt.title('Quantity of Products by Road')
plt.show()
```
