```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,11),
    'product':['pen','pencil',None,'apple','lechi','pencil','pent','apple',None,'pencil'],
    'price':[10,100,20,100,10,None,20,100,None,555],
    'customer':['abudullah','karim','alex','karim','abudullah','jolla','alex','karim',None,'alex'],
    'quantity':[1,2,1,2,1,2,1,2,1,2],
}
df=pd.DataFrame(data)
print("==== raw data ====")
print(df)

```

```python
df['product']=df['product'].fillna('unknown')
df['price']=df['price'].fillna(df['price'].mean())
df['customer']=df['customer'].fillna(method='ffill')
print("==== clean data ====")
print(df)
```

```python
data={
    'order':range(1,11),
    'customer':['abudullah','karim','alex','karim','abudullah','jolla','alex','karim','alex','alex'],
    'product':['pen','pencil','pen','apple','lechi','pencil','pent','apple','pen','pencil'],
    'price':[10,100,20,155,10,None,20,100,None,555],
    'quantity':[1,2,1,2,1,2,1,2,np.nan,2],
}
df=pd.DataFrame(data)
print("==== raw data ====")
print(df)
```

```python
df['price']=df['price'].fillna(df['price'].mean())
df['quantity']=df['quantity'].fillna(df['quantity'].mean())
print("==== clean data ====")
print(df)
```

```python
df['total_sales']=df['price']*df['quantity']
print(df)
```

```python
total_revenue=df['total_sales'].sum()
print("total revenue:",total_revenue)
```

```python
top_customers=df.groupby('customer')['total_sales'].sum().idxmax()
print("top customer:",top_customers)
```

```python
top_product=df.groupby('product')['total_sales'].sum().idxmax()
print("top product:",top_product)
```
