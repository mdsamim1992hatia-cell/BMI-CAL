```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,10001),
    'customer':np.random.choice(['Alex','Blex','Celex',None,None],10000),
    'product':np.random.choice(['APPLE','BALL','CAT',None,None],10000),
    'price':np.random.randint(100,500,10000),
    'quantity':np.random.randint(1,5,10000),
}
df=pd.DataFrame(data)
print("====raw daata====")
print(df)
```

```python
df['customer']=df['customer'].fillna(df['customer'].mode()[0])
df
```

```python
df['product']=df['product'].fillna(df['product'].mode()[0])
df
```

```python
df['total_sales']=df['price']*df['quantity']
df
```

```python
top_customer=df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
top_customer
```

```python
best_product=df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
best_product
```

```python
worst_product=df.groupby('product')['total_sales'].sum().sort_values(ascending=True)
worst_product
```

```python
total_revenue=df['total_sales'].sum()
total_revenue
```

```python
customer_sales=df.groupby('customer')['total_sales'].sum()
customer_sales
```

```python
top_customer=customer_sales.sort_values(ascending=False)
top_customer
```

```python
product_sales=df.groupby('product')['total_sales'].sum()
product_sales
```

```python
avg_order_value=df['total_sales'].mean()
avg_order_value
```

```python
customer_avg=df.groupby('customer')['total_sales'].mean()
customer_avg
```

```python
top_order=df.sort_values(by='total_sales',ascending=False)
top_order
```

```python
df.groupby(['customer','product'])['total_sales'].sum()
```

```python
df.groupby('product').agg({
    'total_sales':'sum',
    'quantity':'sum'
})
```

```python
df.groupby('customer').agg({
    'total_sales':'sum',
    'quantity':'sum'
})
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_customer.index,y=top_customer.values)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Top Customers by Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=best_product.index,y=best_product.values)
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Best Selling Products')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=worst_product.index,y=worst_product.values)
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Worst Selling Products')
plt.show()
```
