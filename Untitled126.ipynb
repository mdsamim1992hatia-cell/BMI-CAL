```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1={
    'key':[1,2,3,4,5],
    'name':['a','b','c','d','e'],
    'age':[20,21,22,23,24]
}
df1=pd.DataFrame(df1)
df1
```

```python
df2={
    'key':[1,2,3,4,5],
    'name':['f','g','h','i','j'],
    'age':[25,26,27,28,29]
}
df2=pd.DataFrame(df2)
df2
```

```python
result=pd.merge(df1,df2,on='key')
result
```

```python
result=pd.merge(df1,df2,on='key',how='left')
result
```

```python
result=pd.merge(df1,df2,on='key',how='right')
result
```

```python
result=pd.merge(df1,df2,on='key',how='outer')
result
```

```python
df=pd.DataFrame({
    'store':['a','b','c','d','e','f','g'],
    'sales':[100,200,300,400,500,600,700],

})
df.groupby('store')['sales'].sum().sort_values(ascending=False)
#
```

```python
df.groupby('store')['sales'].agg(['sum','mean','max','min'])
```

```python
df=pd.DataFrame({
    'customer':['a','b','c','d','e','f','g'],
    'sales':[455,456,45,5555 , 555,555,5555],

})
df.groupby('customer')['sales'].sum().sort_values(ascending=False)
#
```
