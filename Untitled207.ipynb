```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
data={
    'order_id':[1,2,3,4,5,6,7],
    'product_name':['apple','banana','orange',None,'banana',None,'apple'],
    'quantity':[1,2,52,585,55,252,55],
    'price':[10,20,30,40,50,60,70],
    'customer_name':['alice','bob','alice','bob','alice','bob','alice']
}
df=pd.DataFrame(data)
print('===raw data===')
print(df)
```

```python
df['product_name']=df['product_name'].fillna('unknown')
df['quantity']=df['quantity'].astype(int)
df['price']=df['price'].astype(int)
df['totall_sales']=df['quantity']*df['price']
print('===clened data===')
print(df)
```

```python
customer_sales=df.groupby('customer_name')['totall_sales'].sum().sort_values(ascending=False)
product_sales=df.groupby('product_name')['totall_sales'].sum().sort_values(ascending=False)
high_value_orders=df[df['totall_sales']>300]
print(customer_sales)
print("\n=====PTODUCT WISE SALES=====")
print(product_sales)
print("\n=====HIGH VALUE ORDERS=====")
print(high_value_orders)
```

```python
plt.figure()
sns.barplot(x=customer_sales.index,y=customer_sales.values)
plt.title('customer wise sales')
plt.xlabel('customer')
plt.ylabel('sales')
plt.show()
plt.figure()

```

```python
sns.barplot(x=product_sales.index,y=product_sales.values)
plt.title('product wise sales')
plt.xlabel('product')
plt.ylabel('sales')
plt.show()
```
