```python
df1=pd.DataFrame({
    'key':['A','B','C','D'],
    'a1':[1,2,3,4],
    'a2':[5,6,7,8]
})
df1
```

```python
df2=pd.DataFrame({
    'key':['A','B','C','k'],
    'b1':[1,2,3,4,],
    'b2':[5,6,7,8,],
})
df2
```

```python
pd.merge(df1,df2,on='key',how='inner')[['key','a1','b1']]
```

```python
pd.merge(df1,df2,on='key',how='left')[['key','a1','b1']]
```

```python
pd.merge(df1,df2,on='key',how='right')[['key','a1','b1']]
```

```python
pd.merge(df1,df2,on='key',how='outer')[['key','a1','b1']]
```
