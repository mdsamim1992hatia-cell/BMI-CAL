```python
plt.figure(figsize=(8,6))
sns.barplot(x='customer_id',y='total_price',data=df)
plt.title('Total Price by Customer')
plt.xlabel('Customer')
plt.ylabel('Total Price')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.barplot(x='customer_id',y='quantity',data=df)
plt.title('Quantity by Customer')
plt.xlabel('Customer')
plt.ylabel('Quantity')
plt.show()
```
