```python
import pandas as pd
import numpy as np
```

```python
data={
    'order_id':range(1001,1011),
    'country':['India','USA','Japan','China','India','USA','Japan','China','India','USA'],
    'category':['Electronics','Clothing','Books',np.nan,'Clothing','Books','Electronics',np.nan,'Books','Electronics'],
    'sales':[1000,2000,3000,4000,5000,6000,np.nan,np.nan,444,555],
    'profit':[100,200,30,40,50,60,70,80,90,100],
    'order_date':pd.date_range(start='2022-01-01',periods=10,)
}
df=pd.DataFrame(data)
df
```

```python
df.isna().sum()
```

```python
df['category'].fillna('Unknown',inplace=True)
```

```python
df['sales'].fillna(df['sales'].median(),inplace=True)
```
