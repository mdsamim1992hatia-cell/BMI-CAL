```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'name':['A','B','C','D','E'],
    'age':[20,25,None,None,40],
    'salary':[10000,20000,None,None,5000],
    'gender':['M','F','M','F','clib'],
    'department':['IT','HR','ass.manger','HR','manager'],
}
df=pd.DataFrame(data)
print("===raw data===")
print(df)
```

```python
df.isnull().sum()
```

```python
df['age']=df['age'].fillna(df['age'].mean())
df['salary']=df['salary'].fillna(df['salary'].mean())
print("===after fillna===")
print(df)
```

```python
dep_avg_salary=df.groupby('department')['salary'].mean()
print(dep_avg_salary)
```

```python
df['new']=df['department'].map(dep_avg_salary)
print(df)
```

```python
df.loc[df['salary']<df['new'],'salary']=df['new']
print(df)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='department',y='salary',data=df)
plt.title('department vs salary')
plt.show()
```

```python
df.groupby('gender')['salary'].mean().sort_values(ascending=False)
plt.figure(figsize=(10,6))
sns.barplot(x='gender',y='salary',data=df)
plt.title('gender vs salary')
plt.show()
```
