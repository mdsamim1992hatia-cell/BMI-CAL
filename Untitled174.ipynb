```python
import sqlite3
import pandas as pd
```

```python
conn=sqlite3.connect('next.db')
cursor=conn.cursor()
```

```python
cursor.execute("""
create table if not exists employees(
  emp_id int primary key,
  name text,
  department text,
  salary integer,
  age integer
)
""")
conn.commit()
```

```python
data=[
    (1,'sgggamim','sales',5000,28),
    (2,'sattmim','sales',50550,28),
    (3,'sahyhmim','sales',50988900,28),
    (4,'sajjjmim','sales',5555000,28),
    (5,'samihhm','sales',500660,28),
    (6,'sani','sales',500440,28),
]
cursor.executemany("""
insert into employees values (?,?,?,?,?)
""",data)
conn.commit()
```
