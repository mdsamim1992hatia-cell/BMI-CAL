```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'oredr':range(1,1000001),
    'customer':np.random.choice(['customer1','customer2','customer3',None,None,None],1000000),
    'product':np.random.choice(['product1','product2','product3',None,None],1000000),
    'quantity':np.random.randint(1,10,1000000),
    'price':np.random.randint(10,100,1000000),
}
df=pd.DataFrame(data)
print('raw data')
print(df)
```

```python
report=df.groupby('customer').agg({'total_sales':'sum','quantity':'sum'})
report
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=report.index,y=report['total_sales'])
plt.title('total sales by customer')
plt.xlabel('customer')
plt.ylabel('total sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=report.index,y=report['quantity'])
plt.title('total quantity by customer')
plt.xlabel('customer')
plt.ylabel('total quantity')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x=report.index,y=report['total_sales'])
plt.title('total sales by customer')
plt.xlabel('customer')
plt.ylabel('total sales')
plt.show()
```
