```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'oredr_id':range(101,109),
    'store':['dhaka',None,'khulna','dhaka','cumilla','khulna','khulna',None],
    'sales':[555,444,np.nan,None,25252,None,5445,54],
    'quntity':[4,5,6,55,555,855,np.nan,np.nan],
    'date':pd.date_range('2022-01-01',periods=8)

}
df=pd.DataFrame(data)
df
```

```python
df.isna().sum()
```

```python
df['sales'].fillna(df['sales'].mean(),inplace=True)
df
```

```python
df['quntity'].fillna(df['quntity'].mean(),inplace=True)
df
```

```python
df['date']=pd.to_datetime(df['date'])
df
```

```python
df.isna().sum()
```

```python
df['store'].fillna(df['store'].mode()[0],inplace=True)
df
```

```python
df.isna().sum()
```

```python
store_totall_sales=df.groupby('store')['sales'].sum().sort_values(ascending=False)
store_totall_sales
```

```python
store_avg_sales=df.groupby('store')['sales'].mean()
store_avg_sales
```

```python
daily_sales=df.groupby('date')['sales'].sum().sort_values(ascending=False)
daily_sales
```

```python
top_3_orders=df.sort_values('sales',ascending=False).head(3)
top_3_orders
```
