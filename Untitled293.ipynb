```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range (1,100001),
    'customer':np.random.choice(['Alec','Bello',None,'D'],100000),
    'product':np.random.choice(['XOO','YOOO','ZPPP',None],100000),
    'quantity':np.random.randint(1,10,100000),
    'sales':np.random.randint(100,1000,100000)

}
df=pd.DataFrame(data)
df
```

```python
df['customer']=df['customer'].fillna('D')
df.isna().sum()
```

```python
df['product']=df['product'].fillna('ZPPP')
df.isna().sum()
```

```python
df['total_sales']=df['quantity']*df['sales']
df
```

```python
totall_revenue=df['total_sales'].sum()
totall_revenue
```

```python
top_customer=df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
top_customer
```

```python
lowest_customer=df.groupby('customer')['total_sales'].sum().sort_values(ascending=True)
lowest_customer
```

```python
df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_customer.index,y=top_customer.values)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Top Customers')
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=lowest_customer.index,y=lowest_customer.values)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Lowest Customers')
```
