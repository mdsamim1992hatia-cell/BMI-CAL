```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order_id':range(101,111),
    'country':['usa','india','ban','canada','india',None,'canada','italy',None,'italy'],
    'catagory':['tech','food','tech','food','food','food','tech','food','tech','food'],
    'sales':[None,None,555,55,55,8585,55,3636,220,221],
    'profit':[None,None,555,5545,6565,68685,665,3636,220,221],
    'order_date':pd.date_range(start='2022-01-01',periods=10,freq='D')
}
df=pd.DataFrame(data)
print("----raw dat---")
print(df)
```

```python
df['country'].fillna('unknown',inplace=True)
print(df)
```

```python
df['sales'].fillna(df['sales'].mean(),inplace=True)
print(df)
```

```python
df['country'].fillna(method='ffill',inplace=True)
print(df)
```

```python
df=df.dropna()
print(df)
```

```python
df['total_sales']=df['sales']*df['profit']
print(df)
```

```python
df['sales']=df['sales'].astype(int)
print(df)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='country',y='total_sales',data=df)
plt.xlabel('country')
plt.ylabel('total_sales')
plt.title('total sales by country')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='catagory',y='total_sales',data=df)
plt.xlabel('catagory')
plt.ylabel('total_sales')
plt.title('total sales by catagory')
plt.show()
```

```python
plt.figure  (figsize=(10,6))
sns.barplot(x='order_date',y='total_sales',data=df)
plt.xlabel('order_date')
plt.ylabel('total_sales')
plt.title('total sales by order_date')
```
