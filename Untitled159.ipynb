```python
import  numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.arange(20)
y=x**2
print(x,y)
```

```python
plt.scatter(x,y)
plt.plot(x,y)
plt.plot(x,y,'r')
plt.xlabel('x')
plt.ylabel('y')
plt.title('plot')
plt.xticks(range(20))
plt.yticks(range(0,500,40))
plt.show()


```

```python
fig=plt.figure(figsize=(10,10))
axes=fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x,y)
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('plot')
axes.set_xticks(range(20))
axes.set_yticks(range(0,500,40))
```

```python
fig,axes=plt.subplots(2,2,sharex=True,sharey=True,figsize=(10,10))
axes[0,0].plot(x,y,'r-*')
axes[0,0].set_xlabel('x')
axes[0,0].set_ylabel('y')
axes[0,0].set_title('plot')
axes[1,1].scatter(x,y)
```

```python
fig3=plt.figure(figsize=(10,10))
axes_main=fig3.add_axes([0.1,0.1,0.8,0.8])
axes_insert=fig3.add_axes([0.2,0.5,0.4,0.3])
axes_main.plot(x,y,'r-*')
axes_main.set_xlabel('x')
axes_main.set_ylabel('y')
axes_main.set_title('plot')
axes_insert.scatter(x,y)
axes_insert.set_xlabel('x')
axes_insert.set_ylabel('y')
axes_insert.set_title('plot')
```

```python
print(x)
print(x**2)
print(x**3)
fig=plt.figure()
axes=fig.add_axes([0.1,0.1,0.8,0.8])
axes.plot(x,x**2,label='x**2')
axes.plot(x,x**3,label='x**3')
axes.legend(loc=2)
plt.show()
```

```python
fig,axes=plt.subplots()
axes.plot(x,x**2,'r-*')
axes.plot(x,x**3,'b-o')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('plot')
axes.grid(True)
fig.tight_layout()
plt.show()

```
