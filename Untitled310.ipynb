```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
#
```

```python
data={
    'order':range(1,100001),
    'customer':np.random.choice(['c1','c2','c3','c4','c5','c6','c7','c8','c9','c10',None,None,None],size=100000),
    'date':pd.date_range(start='2023-01-01',periods=100000,freq='h'),
    'product':np.random.choice(['p1','p2','p3','p4','p5','p6','p7','p8','p9','p10',None,None],size=100000),
    'quantity':np.random.randint(1,10,size=100000),
    'price':np.random.uniform(10,100,size=100000),
}
df=pd.DataFrame(data)
print(df.head())
```

```python
df.isna().sum()
```

```python
df.isna().sum().sum()
```

```python
df['customer']=df['customer'].fillna(df['customer'].mode()[0])
df['product']=df['product'].fillna(df['product'].mode()[0])
print(df.head())
```

```python
df.isna().sum()
```

```python
df['customer']=df['customer'].fillna(df['customer'].mode()[0])
df['product']=df['product'].fillna(df['product'].mode()[0])
print(df.head())
```

```python
df['total_sales']=df['quantity']*df['price']
print(df.head())
```

```python
df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
```

```python
df['date']=pd.to_datetime(df['date'])
df['month']=df['date'].dt.month
df['year']=df['date'].dt.year
print(df.head())
```

```python
monthly_sales=df.groupby(['year','month'])['total_sales'].sum()
print(monthly_sales)
```

```python
sns.barplot(x=monthly_sales.index.get_level_values(0),y=monthly_sales.values,hue=monthly_sales.index.get_level_values(1))
plt.show()
```
