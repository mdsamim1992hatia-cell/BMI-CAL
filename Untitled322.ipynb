```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
data={
    'order':range (1,10001),
    'customer':np.random.choice(['cust1','cust2','cust3',None,None],size=10000),
    'product':np.random.choice(['item1','item2','item3',None],size=10000),
    'quantity':np.random.randint(1,10,size=10000),
    'price':np.random.randint(10,100,size=10000),

}
df=pd.DataFrame(data)
df
#
```

```python
df.loc[2:5]
```

```python
df.info()
```

```python
df.describe()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df.notna().sum()
```

```python
df.notna().sum().sort_values(ascending=False)
```

```python
df['order']=df['order'].astype('category')
df['customer']=df['customer'].astype('category')
df['product']=df['product'].astype('category')
df
```

```python
df['customer']=df['customer'].fillna(df['customer'].mode()[0])

df
```

```python
df['product']=df['product'].fillna(df['product'].mode()[0])
df
```

```python
df['total_sales']=df['quantity']*df['price']
df
```

```python
df.groupby('customer')['total_sales'].sum()
```

```python
df.groupby('product')['total_sales'].sum()
```

```python
df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby(['product','customer'])['total_sales'].sum()
```

```python
df.groupby(['product','customer'])['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('product').agg({
    'total_sales':'sum',
    'quantity':'sum'
})
```

```python
df.groupby('product')['price'].agg(['min','max','mean'])
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='total_sales',data=df)
plt.xticks(rotation=90)
plt.show()
```
