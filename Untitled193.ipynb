```python
import pandas as pd
```

```python
data={
    'order_id':[1,2,3,4,5],
    'product':['pen','notebook','pen','pencil','notebook'],
    'price':[210,50,50,5,60],
    'quantity':[5,8,6,9,6],
    'customer':['alice','bob','choom','alice','bob']
}
df=pd.DataFrame(data)
df
```

```python
df['totall']=df['price']*df['quantity']
df
```

```python
df[['customer','totall']]
```

```python
df[df['customer']=='alice']
```

```python
product_sales=df.groupby('product')['totall'].sum()
product_sales
```

```python
product_sales=df.groupby('product')['quantity'].sum()
product_sales
```

```python
least_sold=product_sales.idxmin()
least_sold
```

```python
least_sold=product_sales.idxmax()
least_sold
```

```python
np.average=product_sales.mean()
np.average
```

```python
product_sales.sort_values(ascending=False)
product_sales.nlargest(n=3)
```

```python
df.loc[0,'product']
```

```python
df.iloc[0,2]
```

```python
df.isnull().sum()
```

```python
df.fillna(value=0,inplace=True)
df
```

```python
df.groupby('customer').agg({
    'totall':'sum',
    'quantity':'mean'
})
```

```python
import seaborn as sns
```

```python
sns.barplot(x=product_sales.index,y=product_sales.values)
```
