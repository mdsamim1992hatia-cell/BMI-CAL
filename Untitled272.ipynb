```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,1000001),
    'product':np.random.choice(['Apple','Benana','mango',None,None],1000000),
    'price':np.random.randint(10,100,1000000),
    'customer':np.random.choice(['Arer','Bellel','Cwer','Dqwwe',None,None],1000000),
    'Quantity':np.random.randint(1,10,1000000),
}
df=pd.DataFrame(data)
print('======raw data======')
print(df)
```

```python
df['customer']=df['customer'].fillna('Unknown')
df['product']=df['product'].fillna('Unknown')
print('======clean  data======')
print(df)
```

```python
total_revenue=df['price'].sum()
print('Total Revenue:',total_revenue)
```

```python
df['total_sales']=df['price']*df['Quantity']
df
```

```python
total_profit=df['total_sales'].sum()
print('Total Profit:',total_profit)
```

```python
product_sales=df.groupby('product')['total_sales'].sum()
print('Product Sales:')
print(product_sales)
```

```python
customer_sales=df.groupby('customer')['total_sales'].sum()
print('Customer Sales:')
print(customer_sales)
```

```python
oredr_avg=df['total_sales'].mean()
print('Order Average:',oredr_avg)
```

```python
top_customer=customer_sales.idxmax()
print('Top Customer:',top_customer)
```

```python
top_product=product_sales.idxmax()
print('Top Product:',top_product)
```

```python
df.groupby(['customer','product'])['total_sales'].sum()
```

```python
df.groupby(['customer','product'])['total_sales'].sum().unstack(fill_value=0)
```

```python
df.groupby('product').agg({'total_sales':'sum','Quantity':'sum'})
```

```python
df.groupby('customer').agg({'total_sales':'sum','Quantity':'sum'})
df.groupby('customer').agg({'total_sales':'sum','Quantity':'sum'}).reset_index()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=product_sales.index,y=product_sales.values)
plt.title('Product Sales')
plt.xlabel('Product')
plt.ylabel('Total Sales')
```
