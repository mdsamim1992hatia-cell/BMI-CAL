```python
data={
    'order_id':range(11,21),
    'customer':np.random.choice(['A','B','C',None],10),
    'product':np.random.choice(['X','Y',None],10),
    'quantity':np.random.randint(1,10,10),
    'price':np.random.randint(10,100,10),
}
df=pd.DataFrame(data)
print(df)
```

```python
df.isnull().sum()
```

```python
df.isna().sum()
```

```python
df['customer']=df['customer'].fillna('Unknown')
df['product']=df['product'].fillna('Unknown')
print(df)
```

```python
df['total_sales']=df['quantity']*df['price']
print(df)
```

```python
total_revenue=df['total_sales'].sum()
print('Total Revenue:',total_revenue)
```

```python
customer_sales=df.groupby('customer')['total_sales'].sum()
print(customer_sales)
```

```python
top_customers=customer_sales.nlargest(3)
print(top_customers)
```

```python
product_sales=df.groupby('product')['total_sales'].sum()
print(product_sales)
```

```python
top_products=product_sales.nlargest(2)
print(top_products)
```

```python
avarge_sales=df['total_sales'].mean()
print('Avarge Sales:',avarge_sales)
```

```python
avarge_order_value=df.groupby('order_id')['total_sales'].mean()
print(avarge_order_value)
```

```python
customer_avg=df.groupby('customer')['total_sales'].mean().sort_values(ascending=False)
print(customer_avg)
```

```python
top_order=df.groupby('order_id')['total_sales'].sum().nlargest(5)
print(top_order)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='total_sales',data=df)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Total Sales by Customer')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='product',y='total_sales',data=df)
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Total Sales by Product')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='order_id',y='total_sales',data=df)
plt.xlabel('Order ID')
plt.ylabel('Total Sales')
plt.title('Total Sales by Order ID')
plt.show()
```
