```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'oreder_id':[101,102,103,104,105,106,107],
    'customer_id':['a','b','c','a','b','c','c'],
    'age':[25,np.nan,35,25,30,np.nan,30],
    'sales':[200,450,np.nan,300,150,400,400],
    'citi':['dhaka','chittagong','khulna','dhaka','chittagong','khulna','khulna']
}

df=pd.DataFrame(data)
df
```
