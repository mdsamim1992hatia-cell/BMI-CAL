```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order_id':range(1,101),
    'customer':np.random.choice(['ami','tumi',None,'SEAW'],size=100),
    'product':np.random.choice(['laptop','mobile','tablet',None],size=100),
    'quantity':np.random.randint(1,5,size=100),
    'price':np.random.randint(100,1000,size=100),
}
df=pd.DataFrame(data)
print("===RAW DATA===")
print(df)
```

```python
df['customer']=df['customer'].fillna('unknown')
df['product']=df['product'].fillna('unknown')
print(df)
```

```python
df.dropna(inplace=True)
print(df)
```

```python
df['quantity']=df['quantity'].astype(int)
df['price']=df['price'].astype(int)
print(df)
```

```python
df['total_sales']=df['quantity']*df['price']
print(df)
```

```python
top_customer=df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
print(top_customer)
```

```python
top_product=df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
print(top_product)
```

```python
avg_sales=df['total_sales'].mean()
print(avg_sales)
```

```python
avg_order=df.groupby('order_id')['total_sales'].sum().mean()
print(avg_order)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_customer.index,y=top_customer.values)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Top Customers')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_product.index,y=top_product.values)
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Top Products')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['total_sales'],bins=20,kde=True)
plt.xlabel('Total Sales')
plt.ylabel('Frequency')
plt.title('Distribution of Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='quantity',y='price',data=df)
plt.xlabel('Quantity')
plt.ylabel('Price')
plt.title('Quantity vs Price')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='customer',y='total_sales',data=df)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Total Sales by Customer')
plt.xticks(rotation=90)
plt.show()
```
