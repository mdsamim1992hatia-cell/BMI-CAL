```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'a':[1,2,3,4,5,None,None,None],
    'b':[6,7,8,9,10,None,None,None],
    'C':[11,12,13,14,15,None,None,None],
    'D':[16,17,18,19,20,None,None,None],
    'E':[21,22,23,24,25,None,None,None],
    'F':[26,27,28,29,30,None,None,None],
    'G':[31,32,33,34,35,None,None,None],
    'H':[36,37,38,39,40,None,None,None],
    'I':[41,42,43,44,45,None,None,None],

}
df=pd.DataFrame(data)
print("====raw data===")
print(df)
```

```python
df['D']=df['D'].fillna(df['D'].mean())
df['E']=df['E'].fillna(df['E'].mean())
df['F']=df['F'].fillna(df['F'].mean())
df['G']=df['G'].fillna(df['G'].mean())
df['H']=df['H'].fillna(df['H'].mean())
df['I']=df['I'].fillna(df['I'].mean())
df['a']=df['a'].fillna(df['a'].mean())
df['b']=df['b'].fillna(df['b'].mean())
df['C']=df['C'].fillna(df['C'].mean())
print("====clean data===")
print(df)
```

```python
df['new_column']=df['a']+df['b']
df
```

```python
df.drop('new_column',axis=1,inplace=True)
df
```
