```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order_id':[1,2,None,None,5],
    'country':['uk',None,'ban','italy',None],
    'catagory':['A',None,'C','D',None],
    'sales':[None,200,300,400,None],
    'profit':[10,None,30,None,50],
    'oredr_date':pd.date_range('2025-01-01','2025-01-5')
}
df=pd.DataFrame(data)
df
```

```python
df['country'].fillna('zambia',inplace=True)
df
```

```python
df['sales'].fillna(df['sales'].median(),inplace=True)
df
```

```python
df['order_id'].fillna(df['order_id'].median(),inplace=True)
df
```

```python
df['catagory'].fillna(df['catagory'].mode()[0],inplace=True)
df
```

```python
df['profit'].fillna(df['profit'].median(),inplace=True)
df
```
