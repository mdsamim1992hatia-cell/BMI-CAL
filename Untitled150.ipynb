```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x=np.arange(20)
y=x**2
print(x,y)
fig,axes=plt.subplots(2,2,sharex=True,sharey=True,figsize=(8,5))
plt.tight_layout()
axes[0,0].plot(x,y,'g-*')
axes[0,0].set_xlabel('x title')
axes[0][0].set_ylabel('y_title')
axes[0,0].set_title('graph')
axes[1,1].scatter(x,y);
```

```python
fig,axes=plt.subplots(2,2,sharex=True,sharey=True,figsize=(8,5))
plt.tight_layout()
axes[0,0].plot(x,y,'y-*')
axes[0,0].set_xlabel('x axis title')
axes[0][0].set_ylabel('y axis title')
axes[0,0].set_title('graph')
axes[1,1].scatter(x,y);
```

```python
fig,axes=plt.subplots(3,2,sharex=True,sharey=True,figsize=(6,4))
plt.tight_layout()
axes[0,0].plot(x,y,'r-*')
axes[0,1].scatter(x,y)
axes[1,0].plot(x,y,'g-*')
axes[1,1].scatter(x,y);
```

```python
fig,axes=plt.subplots(3,3,sharex=True,sharey=True,figsize=(8,5))
plt.tight_layout()
axes[0,0].plot(x,y,'r-*')
axes[0,1].scatter(x,y)
axes[1,0].plot(x,y,'g-*')
axes[1,1].scatter(x,y)
axes[2,0].plot(x,y,'b-*')
axes[2,1].scatter(x,y)
axes[2,2].scatter(x,y)
axes[0,2].scatter(x,y)
axes[1,2].scatter(x,y)

plt.show()
```

```python
fig3=plt.figure(figsize=(8,5))
axes_main=fig3.add_axes([0,0,1,1])
axes_insert=fig3.add_axes([.50,.50,.6,.5])
axes_main.plot(x,y,'r-*')
axes_insert.scatter(x,y)
plt.show()
```

```python
fig3=plt.figure(figsize=(6,4))
axes_main=fig3.add_axes([0,0,1,1])
axes_insert=fig3.add_axes([.50,.50,.4,.50])
axes_main.plot(x,y,'r-*')
axes_insert.scatter(x,y)
plt.show()
```
