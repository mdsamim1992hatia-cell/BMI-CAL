```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order_id':range(1001,1011),
    'customer_id':[1,2,3,4,np.nan,6,7,np.nan,9,10],
    'customer_name':['tm','tm1','tm2','tm3','tm4',np.nan,'tm6','tm7','tm8','tm9'],
    'sales':[100,np.nan,300,400,500,600,np.nan,np.nan,np.nan,np.nan],

}
df=pd.DataFrame(data)
df
```

```python
df.isna().sum()
```

```python
df['customer_id'].fillna('unknown',inplace=True)
```

```python
df['customer_name'].fillna('unknown',inplace=True)
```

```python
df['sales'].fillna(df['sales'].median(),inplace=True)
```
