```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,1001),
    'customer':np.random.choice(['cust1','cust2',None,None,None],1000),
    'product':np.random.choice(['prod1','prod2',None,None],1000),
    'quantity':np.random.randint(1,10,1000),
    'price':np.random.randint(10,100,1000)
}
df=pd.DataFrame(data)
df
```

```python
df['total_price']=df['quantity']*df['price']
df
```

```python
df[['customer','price']]
```

```python
df[df['customer']]='cust1'
df
```
