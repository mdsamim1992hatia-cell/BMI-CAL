```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,51),
    'customer':np.random.choice(['alex','anex','ases',None],50),
    'product':np.random.choice(['apple','banana','assaim',None],50),
    'quantity':np.random.randint(1,10,50),
    'sales':np.random.randint(100,1000,50)
}
df=pd.DataFrame(data)
print("===RAW DATA ====")
print(df)
```

```python
df['customer']=df['customer'].fillna('unknown')
df['product']=df['product'].fillna('unknown')
print("===CLEAN DATA====")
print(df)
```

```python
top_customer=df.groupby('customer')['sales'].sum().sort_values(ascending=False)
top_customer
```

```python
top_product=df.groupby('product')['sales'].sum().sort_values(ascending=False)
top_product
```

```python
top_sales=df.groupby(['customer','product'])['sales'].sum().sort_values(ascending=False)
top_sales
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_customer.index,y=top_customer.values)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Top Customer')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_product.index,y=top_product.values)
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Top Product')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_sales,y=top_sales.values)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Top Sales')
plt.show()
```
