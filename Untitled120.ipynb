```python
df1={
    'key':[1,2,3,4,5],
    'y1':[11,12,None,15,16],
    'z1':[21,None,23,24,25]
}
df1=pd.DataFrame(df1)
df1
```

```python
df2={
    'key':[1,2,3,55,66],
    'x1':[11,100,44,655,555],
    'z1':[21,22,23,24,25]
}
df2=pd.DataFrame(df2)
df2
```

```python
result=pd.merge(df1,df2,how='inner',on='key')
result
```

```python
result=pd.merge(df1,df2,how='outer',on='key')[['key','x1','y1']]
result
```

```python
result=pd.merge(df1,df2,how='left',on='key')[['x1','y1']]
result
```

```python
result=pd.merge(df1,df2,how='right',on='key')[['x1','y1']]
result
```
