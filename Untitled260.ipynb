```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,10001),
    'price':np.random.randint(10,100,10000),
    'product':np.random.choice(['APPLE','BANANA','CALLM',None,None],10000),
    'customer':np.random.choice(['can','alex',None,None],10000),
    'quantity':np.random.randint(1,10,10000),

}
df=pd.DataFrame(data)
print("====raw data====")
print(df)
```

```python
df['product']=df['product'].fillna('None')
df
```

```python
df['customer']=df['customer'].fillna('None')
df
```

```python
df['price']=df['price'].fillna(df['price'].mean())
df
```

```python
df['total_sales']=df['price']*df['quantity']
df
```

```python
total_revenue=df['total_sales'].sum()
print(total_revenue)
```

```python
customer_sales=df.groupby('customer')['total_sales'].sum()
print(customer_sales)
```

```python
top_customer=customer_sales.idxmax()
print(top_customer)
```

```python
lower_customer=customer_sales.idxmin()
print(lower_customer)
```

```python
product_sales=df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
print(product_sales)
```

```python
avarge_order=df['total_sales'].mean()
print(avarge_order)
```

```python
customer_avg=df.groupby('customer')['total_sales'].mean()
print(customer_avg)
```

```python
top_orders=df.nlargest(5,'total_sales')
print(top_orders)
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['total_sales'],bins=20,kde=True)
plt.title('Distribution of Total Sales')
plt.xlabel('Total Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='total_sales',data=df)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Total Sales by Customer')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='product',y='total_sales',data=df)
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Total Sales by Product')
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='price',y='total_sales',data=df)
plt.xlabel('Price')
plt.ylabel('Total Sales')
plt.title('Total Sales vs Price')
plt.show()
```
