```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns



```

```python
data={
    'Name':['Alice','Bob','Claire',None],
    'Age':[25,30,27,None],
    'Gender':['Female','Male','Female',None],
    'Salary':[50000,60000,55000,None],
    'City':['New York','San Francisco','Los Angeles',None],
    'Country':['USA','USA','USA',None],
    'Department':['HR','IT','Finance',None],
    'quantity':[11,55,55,558,]


}
df=pd.DataFrame(data)
print(df)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna()
```

```python
df.isna().sum().sum()
```

```python
df['Age']=df['Age'].fillna('unknown')
df
```

```python
df['City']=df['City'].fillna(df['City'].mode()[0])
df
```

```python
df['Department']=df['Department'].fillna('un known')
df
```

```python
df['Name']=df['Name'].fillna('kew')
df
```

```python
df['Salary']=df['Salary'].fillna(df['Salary'].mean())
df
```

```python
df['Gender']=df['Gender'].fillna('un known')
df
```

```python
df['Country']=df['Country'].fillna('un known')
df
```

```python
df['total_sales']=df['quantity']*df['Salary']
df
```

```python
df[df['Salary']<55500]
```

```python
df.drop_duplicates()
```

```python
df.groupby('Gender')['total_sales'].mean()
```

```python
df.groupby('Name')['total_sales'].mean()
```

```python
df.groupby('Department')['total_sales'].mean()
```
