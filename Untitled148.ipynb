```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x=np.arange(10)
y=x**2
print(x,y)
plt.scatter(x,y)
plt.plot(x,y,'r-*')
plt.xlabel('x axis title')
plt.ylabel('y axis title')
plt.title('graph title')
plt.xticks(range(0,11))
plt.yticks(range(0,101,10))
plt.show();
```

```python
fig=plt.figure(figsize=(8,5))
axes=fig.add_axes([0,0,1,1])
axes.plot(x,y,'g-*')
axes.set_xlabel('x label')
axes.set_ylabel('y label')
axes.set_xticks(range(0,20))
axes.set_yticks(range(0,101,10))
axes.set_title('graph_chart');
```

```python
fig,axes=plt.subplots(2,2,sharex=True,sharey=True,figsize=(8,5))
plt.tight_layout()
axes[0][0].plot(x,y,'r-*')
axes[0][0].set_xlabel('x label')
axes[0][0].set_ylabel('ylabel')
axes[0][0].set_title('graph_demo')
axes[0][1].scatter(x,y)
```

```python
fig,axes=plt.subplots(2,3,sharex=True,sharey=True,figsize=(7,5))
plt.tight_layout()
axes[0,0].plot(x,y,'r-*')
axes[0,0].set_xlabel('x label')
axes[0,0].set_ylabel('ylabel')
axes[0,0].set_title('graph_demo')
axes[0,1].scatter(x,y)
axes[0,1].plot(x,y,'g-*')
axes[0,1].set_xlabel('x label')
axes[0,1].set_ylabel('ylabel')
axes[0,1].set_title('graph_ok')
axes[1,0].plot(x,y,'g-*')
axes[1,0].set_xlabel('x label')
axes[1,0].set_ylabel('ylabel')
axes[1,0].set_title('graph_ok')
axes[0,1].scatter(x,y)
axes[1,1].plot(x,y,'y-*')
axes[0,2].plot(x,y,'r-*')
axes[1,2].scatter(x,y)
axes[1,2].set_xlabel('x label')

axes[0,0].set_title('graph_ok')
axes[0,1].scatter(x,y)

```
