```python
import numpy as np
import pandas as pd
```

```python
data={
    'order_id':[101,102,103,104,105,106,107,108],
    'store':['A','B',np.nan,'c','A',np.nan,'d','B'],
    'sales':[54554,np.nan,5515,85858,np.nan,5125,222,2122],
    'quantity':[1,2,3,4,5,6,7,8],
    'date':['2025-01-01','2025-01-02','2025-01-03','2025-01-04','2025-01-05','2025-01-06','2025-01-04','2025-01-08']
}
df=pd.DataFrame(data)
df
```

```python
df.isna().sum()
```

```python
df['sales'].fillna(df['sales'].mean(),inplace=True)
df
```

```python
df['quantity'].fillna(df['quantity'].mean(),inplace=True)
df
```

```python
df['date']=pd.to_datetime(df['date'])
df
```

```python
store_totall_sales=df.groupby('store')['sales'].sum().sort_values(ascending=False)
store_totall_sales
```

```python
store_avg_sales=df.groupby('store')['sales'].mean()
store_avg_sales
```

```python
daily_sales=df.groupby('date')['sales'].sum()
daily_sales
```

```python
top_3_orders=df.sort_values('sales',ascending=False).head(3)
top_3_orders
```
