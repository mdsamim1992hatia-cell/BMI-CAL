```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
#
```

```python
data={
    'order':range(1,100001),
    'customer':[np.random.choice(['A','B','C','D',None,None,'E']) for i in range(1,100001)],
    'product':[np.random.choice(['P1','P2','P3',None,None,'P4','P5']) for i in range(1,100001)],
    'quantity':[np.random.randint(1,10) for i in range(1,100001)],
    'price':[np.random.randint(10,100) for i in range(1,100001)],
    'DATE':[pd.date_range('2022-01-01','2023-01-01')[np.random.randint(0,365)] for i in range(1,100001)]
}
df=pd.DataFrame(data)
df.head()




```

```python
df.describe()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df['customer']=df['customer'].fillna(df['customer'].mode()[0])
df
```

```python
df['product']=df['customer'].fillna(df['product'].mode()[0])
df
```

```python
df['total_sales']=df['quantity']*df['price']
df
```

```python
df['total_sales'].sum()
```

```python
df.groupby('customer')['total_sales'].sum()
```

```python
df.groupby('product')['total_sales'].sum()
```

```python
df['DATE']=pd.to_datetime(df['DATE'])
df
```

```python
df['month']=df['DATE'].dt.month
df
```

```python
monthly_sales=df.groupby('month')['total_sales'].sum()
monthly_sales
```

```python
sns.barplot(x=monthly_sales.index,y=monthly_sales.values)
plt.show()
```
