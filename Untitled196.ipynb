```python
report=df.groupby('customer_id').agg({
    'order_id':'count',
    'price':'sum',
    'quantity':'sum',
    'totall':'sum'
})
report

```

```python
sns.barplot(
    x=report.index,
    y=report['order_id']
)
plt.title('totall sales by customer')
plt.show()
```
