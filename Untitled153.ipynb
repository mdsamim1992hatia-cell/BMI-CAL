```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.arange(20)
y=x**2
print(x,y)
```

```python
plt.scatter(x,y)
plt.plot(x,y,'r-*')
plt.xlabel('x axis title')
plt.ylabel('y axis titel')
plt.title('graph chart')
plt.xticks(range(11))
plt.yticks(range(0,400,50));
```

```python
fig1=plt.figure(figsize=(8,5))
axes=fig1.add_axes([0,0,1,1])
axes.plot(x,y,'r-*')
axes.set_xlabel('x title')
axes.set_ylabel('y lavel')
axes.set_title('title')
axes.set_xticks(range(20))
axes.set_yticks(range(0,400,20))
axes.set_title('title')
```

```python
fig,axes=plt.subplots(2,2,sharex=True,sharey=True,figsize=(8,5))
axes[0,0].plot(x,y,'r-*')
axes[0,0].set_xlabel('x title')
axes[0,0].set_ylabel('y lavel')
axes[0,0].set_title('title')
axes[0,0].set_xticks(range(10))
axes[0,0].set_yticks(range(100,10))
axes[0,0].set_title('title')



```
