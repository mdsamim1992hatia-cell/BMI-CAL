```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,11),
    'customer_id':[None,'B',np.nan,'A',None,'C','A',None,'C','A'],
    'product':['A',None,'C','A','B','C','A',None,'C',None],
    'quantity':[5,3,2,4,6,2,3,5,4,6],
    'price':[10,15,None,18,222,np.nan,np.nan,55,554,6558],

}
df=pd.DataFrame(data)
print("===raw data===")
print(df)
```

```python
print("duplicate rows",df.duplicated().sum())
print("duplicate columns",df.columns.duplicated().sum())
df
```

```python
print(df['customer_id'].unique())
```

```python
print(df['product'].nunique())
```

```python
print(df['price'].unique())
```

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

```python
df['customer_id']=df['customer_id'].fillna('unknown')
df
```

```python
df['customer_id']=df['customer_id'].fillna(method='ffill')
df
```

```python
df['price']=df['price'].fillna(df['price'].mean())
df
```

```python
df['product']=df['product'].fillna(df['product'].mode()[0])
df
```

```python
df[['customer_id','product']]
```

```python
df['total_sales']=df['quantity']*df['price']
df
```

```python
df.groupby('customer_id')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby(['customer_id','product'])['total_sales'].sum()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer_id',y='total_sales',data=df)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='product',y='total_sales',data=df)
plt.xticks(rotation=90)
```
