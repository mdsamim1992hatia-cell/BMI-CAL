```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.DataFrame({
    'store':['A','A','A','A','A','B','B','B','B','B'],
    'sales':[1000,20000,3000,4000,5000,1000,555,888,888,88]
})
df.groupby('store')['sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('store')['sales'].sum().sort_values(ascending=True)
```

```python
df.groupby('store')['sales'].agg(['sum','mean','count','max'])
```

```python
df=pd.DataFrame({
    'customer':['A','B','C','D','E','F','A','B','C','D','E','F'],
    'sales':[555,888,999,777,666,55,555,555,555,8585,252,222,]
})
df.groupby('customer')['sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('customer')['sales'].agg(['sum','mean','min','max'])
```

```python
df=pd.DataFrame({
    'dept':['A','A','A','A','A','B','B','B','B','B'],
    'salary':[100,200,300,400,500,555,55555,444,88,22]
})
df.groupby('dept')['salary'].describe()
```

```python
df=pd.DataFrame({
    'day':['mon','tue','wed','thu','fri','sat','sun'],
    'sales':[44,55,66,77,88,99,100]
})
df.groupby('day')['sales'].sum().sort_values(ascending=False)
```

```python
df=pd.DataFrame({
    'product':['mon','tue','wed','thu','fri','sat','sun'],
    'sales':[45954,4444,444,88,88,54,54]
})
df.groupby('product')['sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('product')['sales'].agg(['sum','mean','min','max'])
```
