```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,1001),
    'product':np.random.choice(['apple','LYCHI',None],1000),
    'quantity':np.random.randint(1,10,1000),
    'price':np.random.randint(10,100,1000),
    'customer':np.random.choice(['alex','cutttom',None],1000),

}
df=pd.DataFrame(data)
print("====raw daat====")
print(df)
```

```python
df['customer']=df['customer'].fillna('unknown')
df['product']=df['product'].fillna('unknown')
print("====clean data====")
print(df)
```

```python
df['total_sales']=df['quantity']*df['price']
print("====total sales====")
print(df)
```

```python
total_reveneue=df['total_sales'].sum()
print("====total revenue====")
print(total_reveneue)
```

```python
customer_sales=df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
print("====customer sales====")
print(customer_sales)
```

```python
top_customer=customer_sales.idxmax()
print("====top customer====")
print(top_customer)
```

```python
product_sales=df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
print("====product sales====")
print(product_sales)
```

```python
top_product=product_sales.idxmax()
print("====top product====")
print(top_product)
```

```python
avg_order_value=df['total_sales'].mean()
print("====average order value====")
print(avg_order_value)
```

```python
customer_avg=df.groupby('customer')['total_sales'].mean().sort_values(ascending=False)
print("====customer average====")
print(customer_avg)
```

```python
top_orders=df.nlargest(5,'total_sales')
print("====top orders====")
print(top_orders)
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=customer_sales.index,y=customer_sales.values)
plt.title('customer sales')
plt.xlabel('customer')
plt.ylabel('total sales')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=product_sales.index,y=product_sales.values)
plt.title('product sales')
plt.xlabel('product')
plt.ylabel('total sales')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x='customer',y='total_sales',data=df)
plt.title('customer sales')
plt.xlabel('customer')
plt.ylabel('total sales')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(x='product',y='total_sales',data=df)
plt.title('product sales')
plt.xlabel('product')
plt.ylabel('total sales')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(df['total_sales'],bins=20)
plt.title('total sales distribution')
plt.xlabel('total sales')
plt.ylabel('frequency')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x='quantity',y='total_sales',data=df)
plt.title('quantity vs total sales')
plt.xlabel('quantity')
plt.ylabel('total sales')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot(x='price',y='total_sales',data=df)
plt.title('price vs total sales')
plt.xlabel('price')
plt.ylabel('total sales')
plt.show()
```
