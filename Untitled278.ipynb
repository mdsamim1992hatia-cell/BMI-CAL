```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
   ' order':range(1,10001),
   'product':np.random.choice(['apple','ball','call','derrm',None,None],10000),
   'quantity':np.random.randint(12,60,10000),
   'customer':np.random.choice(['alex','asex',None,None],10000),
   'PRICE':np.random.randint(200,20000,10000),
}
df=pd.DataFrame(data)
print("======raw data =====")
df
```

```python
df.drop_duplicates(inplace=True)
```

```python
df[['product','customer']]
```

```python
df['product']=df['product'].fillna('none')
df['customer']=df['customer'].fillna('none')
print("====clean data====")
print(df)
```

```python
df['total_sales']=df['quantity']*df['PRICE']
df
```

```python
total_revenue=df['total_sales'].sum()
print("total revenue:",total_revenue)
```

```python
product_sales=df.groupby('product')['total_sales'].sum()
print("product sales:")
print(product_sales)
```

```python
customer_sales=df.groupby('customer')['total_sales'].sum()
print("customer sales:")
print(customer_sales)
```

```python
top_customer=customer_sales.idxmax()
print("top customer:",top_customer)
```

```python
top_product=product_sales.idxmax()
print("top product:",top_product)
```

```python
df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby(['customer','product'])['total_sales'].sum()
```

```python
df.groupby('customer').agg({
    'total_sales':['sum','mean','max','min']
})
```

```python
df.groupby('product').agg({
    'total_sales':['sum','mean','max','min']
})
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=customer_sales.index,y=customer_sales)
plt.xlabel('customer')
plt.ylabel('total sales')
plt.title('customer sales')
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=product_sales.index,y=product_sales)
plt.xlabel('product')
plt.ylabel('total sales')
plt.title('product sales')
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['total_sales'],bins=20,kde=True)
plt.xlabel('total sales')
plt.ylabel('frequency')
plt.title('distribution of total sales')
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='quantity',y='total_sales',data=df)
plt.xlabel('quantity')
plt.ylabel('total sales')
plt.title('quantity vs total sales')
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='product',y='total_sales',data=df)
plt.xlabel('product')
plt.ylabel('total sales')
plt.title('total sales by product')
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='customer',y='total_sales',data=df)
plt.xlabel('customer')
plt.ylabel('total sales')
plt.title('total sales by customer')
```
