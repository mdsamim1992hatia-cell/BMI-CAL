```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':[1,52,53,45,5,6,7],
    'product':['book','pen','pencil','book','pen','pencil','book'],
    'price':[100,5555,52125,51655,257725,2525,2525],
    'customer':['herry','trem','kool','herry',None,'kool',None],
    'quantity':[1,25,3,54,5,685,7]


}
df=pd.DataFrame(data)


print("\n===dataset====")
print(df)
```

```python
df['customer']=df['customer'].fillna('unknown').astype(str)
df['product']=df['product'].astype('category')
df['total_price']=df['price']*df['quantity' ]

df['quantity']=df['quantity'].astype('int64')
print("=====clean dataset=====")
print(df)
```

```python
customer_sales=df.groupby('customer')['total_price'].sum().sort_values(ascending=False)
product_sales=df.groupby('product')['total_price'].sum().sort_values(ascending=False)
high_value_order=df[df['total_price']>300].sort_values('total_price',ascending=False)
print("=====customer sales=====")
print(customer_sales)
print("=====product sales=====")
print(product_sales)
print("=====high value order=====")
print(high_value_order)


```

```python
plt.figure()
sns.barplot(x=customer_sales.index,y=customer_sales.values)
plt.title('customer sales')
plt.xlabel('customer')
plt.ylabel('total price')
plt.show()
```

```python
plt.figure()
sns.barplot(x=product_sales.index,y=product_sales.values)
plt.title('product sales')
plt.xlabel('product')
plt.ylabel('total price')
plt.show()
```

```python
plt.figure()
sns.barplot(x=high_value_order['order'],y=high_value_order['total_price'])
plt.title('high value order')
plt.xlabel('order')
plt.ylabel('total price')
plt.show()
```
