```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
day=np.arange(11)
sale=day**2
print(day,sale)
plt.scatter(day,sale)
```

```python
plt.plot(day,sale)
plt.plot(day,sale,'r-*')
plt.xlabel('day axis title')
plt.ylabel('sale axis title')
plt.title('graph_chart')
plt.show()
```

```python
plt.xticks(day)
plt.yticks(sale)
plt.plot(day,sale);
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tom=np.arange(20)
jerry=tom**2
print(tom,jerry)
plt.scatter(tom,jerry)
plt.plot(tom,jerry,'r-*')
plt.xlabel('tom axis title')
plt.ylabel('jerry axis title')
plt.title('graph')
plt.xticks(range(0,10))
plt.yticks(range(0,400,50));
```

```python
plt.xticks(range(0,11))
plt.yticks(range(0,101,10));
```

```python
fig1=plt.figure(figsize=(8,5))
axes=fig1.add_axes([0,0,1,1])
axes.plot(tom,jerry,'r-*')
axes.set_xlabel(' tom label')
axes.set_ylabel(' jerry label')
axes.set_title('figure title')
axes.set_xticks(range(0,11))
axes.set_yticks(range(0,400,50))
fig1.savefig('figure.png');
```

```python
fig2=plt.figure(figsize=(6,3))
axes=fig2.add_axes([0,0,0.50,0.50])
axes.plot(tom,jerry,'r-*')
axes.set_xlabel('tom label')
axes.set_ylabel('jerry label')
axes.set_title('figure title')
axes.set_xticks(range(0,11))
axes.set_yticks(range(0,400,50))
fig1.savefig('figure.png');
```

```python
fig3=plt.figure(figsize=(8,4))
axes=fig3.add_axes([0,0,1,1])
axes.plot(tom,jerry,'r-*')
axes.set_xlabel('x lable')
axes.set_ylabel('y lable')
axes.set_title('graph size')
axes.set_xticks(range(22))
axes.set_yticks(range(0,400,50))
plt.show();
```

```python
fig5=plt.figure(figsize=(8,8))
axes=fig5.add_axes([0,0,1,1])
axes.plot(tom,jerry,'g-*')
axes.set_xlabel('x axes lable')
axes.set_ylabel('y axes lable')
axes.set_title('new_graph')


axes.set_xticks(range(0,10))
axes.set_yticks(range(0,400,50))
fig5.savefig('figure.png');
```
