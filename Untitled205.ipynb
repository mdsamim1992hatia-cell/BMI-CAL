```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order_id':[1,2,3,4,5],
    'customer_id':['alice','non','alice','bob','alice'],
    'date':['2-2-05','3-3-07','4-4-06','5-5-07','6-6-08'],
    'product_id':['apple','banana','lychi','mango','apple'],
    'quantity':[2,3,1,4,2],
    'price':[10.99,15.99,8.99,111,666]
}
df=pd.DataFrame(data)
df
```

```python
df['totall']=df['quantity']*df['price']
df
```

```python
df[['customer_id','product_id']]
```

```python
df.groupby('customer_id')['totall'].sum().sort_values(ascending=False)
```

```python
df.groupby('product_id')['totall'].sum().sort_values(ascending=False)
```

```python
df.groupby('date')['totall'].sum().sort_values(ascending=False)
```

```python
df.groupby('quantity')['totall'].count()
```

```python
df['date']=pd.to_datetime(df['date'])
df['month']=df['date'].dt.month
df['year']=df['date'].dt.year
df['day']=df['date'].dt.day
monthly_sales=df.groupby(['month'])['totall'].sum().sort_values(ascending=False)
monthly_sales
```

```python
sns.barplot(x=monthly_sales.index,y=monthly_sales.values)
plt.show()
```
