```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
#
```

```python
data={
    'order':range(1,10001),
    'customer_id':np.random.choice(1000,10000),
    'product_id':np.random.choice(100,10000),
    'quantity':np.random.randint(1,5,10000),
    'price':np.random.randint(10,100,10000),
}
df=pd.DataFrame(data)
df
```

```python
df.head()
```

```python
df.tail()
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.groupby('customer_id')['price'].sum().sort_values(ascending=False)
```

```python
df['total_sales']=df['quantity']*df['price']
df
```

```python
df.groupby('customer_id')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('product_id')['quantity'].sum().sort_values(ascending=False)
```

```python
df.groupby('product_id')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('order')['total_sales'].sum().sort_values(ascending=False)
```

```python
customer_avg_sales=df.groupby('customer_id')['total_sales'].mean()
customer_avg_sales
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=customer_avg_sales.index,y=customer_avg_sales.values)
plt.xlabel('Customer ID')
plt.ylabel('Average Sales')
plt.title('Average Sales per Customer')
plt.show()
```
