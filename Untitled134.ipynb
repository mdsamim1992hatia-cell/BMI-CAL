```python
import pandas as pd
import numpy as np
```

```python
data={
    'order_id':range(1,11),
    'customer':[None,'b','c','d',None,'a','c','s','m',None],
    'product':['mobile','laptop',None,'mobile','laptop','tv',None,'laptop','tv',None],
    'quantity':[1,None,3,4,5,6,None,8,9,10],
    'price':[1000,1122112,555,None,522525,555,55,None,55,None],
    'sales':[555,45444,None,666,3633,333,55,None,525,55]

}
df=pd.DataFrame(data)
df
```
