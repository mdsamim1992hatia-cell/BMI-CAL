```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,51),
    'customer':np.random.choice(['ales','alex','mina',None],50),
    'price':np.random.randint(10,100,50),
    'quantity':np.random.randint(1,5,50),
    'total':np.random.randint(100,1000,50),
    'product':np.random.choice(['laptop','phone','tablet'],50)
}
df=pd.DataFrame(data)
print(df)
```

```python
df['total_sales']=df['price']*df['quantity']
print(df)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_customers.index,y=top_customers.values)
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.title('Top 5 Customers by Total Sales')
plt.show()
```
