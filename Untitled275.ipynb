```python
df['Price']=df['Price'].fillna(0)
df['product']=df['product'].fillna('no product')
df['customer_id']=df['customer_id'].fillna('no customer')
df['quantity']=df['quantity'].fillna(0)
print("======clean daat====")
print(df)
```

```python
df['total_sales']=df['Price']*df['quantity']
print(df)
```

```python
total_revenue=df['total_sales'].sum()
print("Total Revenue:",total_revenue)
```

```python
customer_sales=df.groupby('customer_id')['total_sales'].sum()
print("Customer Sales:")
print(customer_sales)
```

```python
product_sales=df.groupby('product')['total_sales'].sum()
print("Product Sales:")
print(product_sales)
```

```python
top_customer=customer_sales.idxmax()
print("Top Customer:",top_customer)
```

```python
lower_customer=customer_sales.idxmin()
print("Lower Customer:",lower_customer)
```

```python
avg_order=df['total_sales'].mean()
print("Average Order:",avg_order)
```

```python
avg_sales=df.groupby('customer_id')['total_sales'].mean()
print("Average Sales:")
print(avg_sales)
```

```python
top_order=df.nlargest(3,'total_sales').sort_values(by='order',ascending=False)

print("Top Order:")
print(top_order)
```

```python
df.groupby(['customer_id','product'])['total_sales'].sum()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='product',y='total_sales',data=df)
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.title('Total Sales by Product')
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='order',y='total_sales',data=df)
plt.xlabel('Order')
plt.ylabel('Total Sales')
plt.title('Total Sales by Order')
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer_id',y='quantity',data=df)
plt.xlabel('Customer ID')
plt.ylabel('Total Quantity')
plt.title('Total Quantity by Customer')
```
