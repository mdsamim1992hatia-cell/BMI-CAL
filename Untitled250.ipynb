```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,1001),
    'product':np.random.choice(['Apple','Ball','Cat',None,None],1000),
    'customer':np.random.choice(['alex','butta','mwew',None,None],1000),
    'quantity':np.random.randint(1,5,1000),
    'price':np.random.randint(10,100,1000)
}
df=pd.DataFrame(data)
print("====raw data=====")
print(df)
```

```python
df['customer']=df['customer'].fillna('unknown')
df
```

```python
df['product']=df['product'].fillna('unknown')
df
```

```python
df['price']=df['price'].fillna(df['price'].mean())
df
```

```python
df['total_sales']=df['quantity']*df['price']
df
```

```python
total_revenue=df['total_sales'].sum()
total_revenue
```

```python
customer_sales=df.groupby('customer')['total_sales'].sum()
customer_sales
```

```python
top_customer=customer_sales.idxmax()
top_customer
```

```python
top_customer_sales=customer_sales.max()
top_customer_sales
```

```python
product_sales=df.groupby('product')['total_sales'].sum()
product_sales
```

```python
top_product=product_sales.idxmax()
top_product
```

```python
avarage_order_value=df['total_sales'].mean()
avarage_order_value
```

```python
customer_avg=df.groupby('customer')['total_sales'].mean()
customer_avg
```

```python
top_orders=df.nlargest(5,'total_sales')
top_orders
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='total_sales',data=df)
plt.title('Total Sales by Customer')
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='product',y='total_sales',data=df)
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='quantity',y='total_sales',data=df)
plt.title('Quantity vs Total Sales')
plt.xlabel('Quantity')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='price',y='total_sales',data=df)
plt.title('Price vs Total Sales')
plt.xlabel('Price')
plt.ylabel('Total Sales')
plt.show()
```
