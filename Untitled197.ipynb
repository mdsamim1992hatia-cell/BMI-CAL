```python
import pandas  as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data={
    'order_id':[1,2,3,4,5],
    'product':['pen','pencil','notebook','pen','pencil'],
    'price':[10,20,30,40,50],
    'quantity':[1,2,3,4,5],
    'customer':['lice','ban','alice','ban','oll']
}
df=pd.DataFrame(data)
df
```

```python
df['totall']=df['price']*df['quantity']
df
```

```python
df[['product','price','totall']]
```

```python
df[df['customer']=='b']
```

```python
df[df['order_id']>2]
```

```python
df[df['product']=='pen']
```

```python
df.groupby('customer')['totall'].sum()
```

```python
df.groupby('customer')['totall'].mean()
```

```python
df.groupby('customer')['totall'].sum().nlargest(2)
```

```python
df.groupby('customer')['totall'].sum().nsmallest(2)
```

```python
df.groupby('customer')['quantity'].sum()
```

```python
report=df.groupby('customer')['totall'].sum().nlargest(2)
report
```

```python
report=df.groupby('customer').agg({
    'totall':'sum',
    'quantity':'sum',
})
report
```

```python
sns.barplot(x=report.index,y=report['totall'])
plt.title('totall')
```
