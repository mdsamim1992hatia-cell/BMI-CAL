```python
import pandas as pd
import numpy as np
```

```python
data={
    'order':[1,2,3,45,5],
    'customer':['alice','bob','charlie','bob','alice'],
    'email':['alice@gmail.com',np.nan,'bob@gmail.com',  None,'charlie@gmail.com'],
    'quantity':[1,2,3,4,5],
    'price':[10,10,15,1000,10000],
    'product':['apple','banana',None,'date','elderberry']
}
df=pd.DataFrame(data)

print("orginal data:\n",df)
df=df.drop_duplicates()
df['customer']=df['customer'].str.title()
df['email']=df['email'].fillna('alice@gmail.com')
df['product']=df['product'].fillna('apple')
df['price']=pd.to_numeric(df['price'],errors='coerce')
df=df[df['price']<1000]
print("\ncleaned data\n",df)




```
