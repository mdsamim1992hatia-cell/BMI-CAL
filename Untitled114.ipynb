```python
df1=pd.DataFrame({
    'key':[10,20,30,None,np.nan,np.nan],
    'b1':[10,2000,300,None,np.nan,np],
    'c1':[10,20,3,10,10,40]})
df1
```

```python
df2=pd.DataFrame({
    'key':[10,20,30,None,60],
    'b2':[10,2000,300,None,88],
    'c2':[10,20,3,10,10]})
df2
```

```python
result=pd.concat([df1,df2])
result
```

```python
result=pd.merge(df1,df2,how='inner',on='key')
result

```

```python
df=pd.merge(df1,df2,how='outer',on='key')
df
```

```python
data=pd.DataFrame({
    'store':['a','b','c','d','a',],
    'sales':[100,200,300,400,500]})
df=pd.DataFrame(data)
df
```

```python
df.groupby('store').mean()
```

```python
df.groupby('store')['sales'].agg(['mean','sum','max','min'])
```

```python
df.groupby('store').sum().sort_values('sales',ascending=False)
```
