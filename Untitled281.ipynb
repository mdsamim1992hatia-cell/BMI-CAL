```python
df['customer']=df['customer'].fillna('Unknown')
df
```

```python
df['product']=df['product'].fillna('Unknown')
df
```

```python
df['quantity']=df['quantity'].fillna(df['quantity'].mean())
df
```

```python
df['total_price']=df['quantity']*df['price']
df
```

```python
total_revenue=df['total_price'].sum()
total_revenue
```

```python
top_customer=df.groupby('customer')['total_price'].sum().sort_values(ascending=False)
top_product=df.groupby('product')['total_price'].sum().sort_values(ascending=False)
avg_price=df['price'].mean()
avg_quantity=df['quantity'].mean()
avg_order=df['total_price'].mean()
print("=====clean data====")
print(top_customer)
print(top_product)
print(avg_price)
print(avg_quantity)
print(avg_order)
```
