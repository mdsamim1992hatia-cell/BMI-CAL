```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.arange(20)
y=x**2
print(x,y)
fig1=plt.figure(figsize=(8,5))
axes=fig1.add_axes([0,0,1,1])
axes.plot(x,y,'r-*')
axes.set_xlabel('x lable')
axes.set_ylabel('y lable')
axes.set_title('graph')
axes.set_xticks(range(0,10))
axes.set_yticks(range(0,400,50))
axes.set_title('graph')
```

```python
x1=np.arange(20)
y1=x1**2
print(x1,y1)
fig=plt.figure(figsize=(6,4))
axes=fig.add_axes([0,0,1,1])
axes.plot(x1,y1,'r-*')
axes.set_xlabel('x lable')
axes.set_ylabel('y lable')
axes.set_title('graph')
axes.set_xticks(range(0,20))
axes.set_yticks(range(0,400,50))
fig.savefig('graph.png');
```
