```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'a':[21,55,np.nan,44,np.nan],
    'b':[32,55,554,55,558],
    'c':[np.nan,np.nan,np.nan,np.nan,np.nan],
    'd' :[55,np.nan,544,55,np.nan]
}
df=pd.DataFrame(data)
print("=====REAL DATA=====")
print(df)
```

```python
def totall_nan(df):
    return df.isnull().sum()
totall_nan(df)

```

```python
df_ffill=df.fillna(method='ffill')
print("=====FORWARD FILL=====")
print(df_ffill);
```

```python
df_bfill=df.fillna(method='bfill')
print("=====BACKWARD FILL=====")
print(df_bfill);
```

```python
df_zero=df.fillna(0)
print("=====ZERO FILL=====")
print(df_zero);
```

```python
data={
    'order':range(1,11),
    'customer_id':np.random.randint(101,111,10),
    'item':['item','item','item','item','item','item',None,'item',None,None],
    'quantity':np.random.randint(1,10,10),
    'price':[14414,44,454,44,44,55,55,None,None,None],
}
df=pd.DataFrame(data)
print('=====REAL DATA=====')
print(df)
```

```python
df.isnull().sum().sum()
print('=====MISSING VALUES=====')
print(df)
```

```python
df['customer_id']=df['customer_id'].fillna(method='ffill')
print('=====FORWARD FILL=====')
print(df)
```

```python
df['item']=df['item'].fillna(method='ffill')
print('=====BACKWARD FILL=====')
print(df)
```

```python
df['price']=df['price'].fillna(df['price'].median())
print('=====MEAN FILL=====')
print(df)
```

```python
df_zero=df.fillna(0)
print("=====ZERO FILL=====")
print(df_zero);
```
