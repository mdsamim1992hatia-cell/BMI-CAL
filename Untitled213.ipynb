```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,11),
    'customer_id':np.random.randint(101,111,10),
    'item':['item','item','item','item','item','item',None,'item',None,None],
    'quantity':np.random.randint(1,10,10),
    'price':[14414,None,None,44,44,55,5,44,655,55]

}
df=pd.DataFrame(data)
print('=====REAL DATA=====')
print(df)
```

```python
print('=====MISSING VALUES=====')
print(df.isnull().sum())
```

```python
df['item']=df['item'].fillna(method='ffill')
print('=====FORWARD FILL=====')
print(df)
```

```python
df['price']=df['price'].fillna(df['price'].median())
print('=====MEAN FILL=====')
print(df)
```

```python
df['quantity']=df['quantity'].fillna(df['quantity'].median())
print('=====clean data=====')
print(df)
```

```python
print("\nloc demo:")
print(df.loc[0:3,['customer_id','price']])
```

```python
print("\niloc demo")
print(df.iloc[0:3,[1,4]])
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x='customer_id',y='quantity',data=df)
plt.title('CUSTOMER ID VS QUANTITY')
plt.xlabel('CUSTOMER ID')
plt.ylabel('QUANTITY')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df.select_dtypes(include=np.number).corr(),annot=True)
plt.title("CORRELATION MATRIX")

plt.show()
```

```python
df['totatl_sales']=df['quantity']*df['price']
print('=====TOTAL SALES=====')
print(df)
```

```python
top_customer=df.groupby('customer_id')['totatl_sales'].sum().sort_values(ascending=False)
print('=====TOP CUSTOMER=====')
print(top_customer)
```

```python
median_sales=df['totatl_sales'].median()
print('=====MEDIAN SALES=====')
print(median_sales)
```

```python
high_value_orders=df[df['totatl_sales']>median_sales]
print('=====HIGH VALUE ORDERS=====')
print(high_value_orders)
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x='customer_id',y='totatl_sales',data=df)
plt.title('CUSTOMER ID VS TOTAL SALES')
plt.xlabel('CUSTOMER ID')
plt.ylabel('TOTAL SALES')
plt.show()
```
