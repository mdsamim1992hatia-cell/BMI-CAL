```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,100001),
    'customer':np.random.choice(['akes','bakes','cakes',None],100000),
    'quantity':np.random.randint(1,10,100000),
    'product':np.random.choice(['apple','banana','carrot',None],100000),
    'price':np.random.randint(10,100,100000),
}
df=pd.DataFrame(data)
df.head()
```

```python
df['customer']=df['customer'].fillna('unknown')
df['product']=df['product'].fillna('unknown')
print("clean data")
print(df)
```

```python
df['total_sales']=df['quantity']*df['price']
print(df)
```

```python
top_customer=df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
print(top_customer)
```

```python
top_order=df.groupby('order')['total_sales'].sum().sort_values(ascending=False)
print(top_order)
```

```python
df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=top_customer.index,y=top_customer.values)
plt.xlabel('customer')
plt.ylabel('total_sales')
plt.title('top customer')
```
