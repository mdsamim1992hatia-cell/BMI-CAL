```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)
```

```python
data={
    'order':range(1,51),
    'customer_id':np.random.randint(1,101,50),
    'sales':np.random.randint(100,1000,50),
    'profit':np.random.randint(10,500,50),
    'quantity':np.random.randint(1,10,50),
    'discount':np.random.uniform(0,0.5,50),
}
df=pd.DataFrame(data)
df
```

```python
df.loc[df['sales']>600]
```

```python
df.loc[10,'sales']=None
print(df.loc[10])
```

```python
df.loc[10,'profit']=None
df
```

```python
df['total_sales']=df['sales']*df['quantity']
df
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer_id',y='total_sales',data=df)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer_id',y='profit',data=df)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer_id',y='quantity',data=df)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer_id',y='discount',data=df,hue='quantity')
```
