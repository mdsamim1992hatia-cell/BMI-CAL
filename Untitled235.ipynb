```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)
```

```python
data={
    'order':range(1,11),
    'customer':['alex','bear','rewxe','alex',np.nan,np.nan,'rewxe','alex','bear','rewxe'],
    'product':['product1','product2','product3','product1','product2','product3','product1',None,None,None],
    'quantity':np.random.randint(1,10,10),
    'price':np.random.randint(10,100,10),
}
df=pd.DataFrame(data)
print("====raw data=====")
print(df)
```

```python
df['customer']=df['customer'].fillna('unknown')
df['product']=df['product'].fillna('unknown')
print("====after fillna=====")
print(df)
```

```python
df['total_sales']=df['quantity']*df['price']
df
```

```python
df[df['customer']=='alex']
```

```python
df.loc[df['customer']=='alex']
```

```python
df[['price','quantity']]
```

```python
df[df['price']<30]
```

```python
df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
#
```

```python
df.groupby('customer')['quantity'].sum().sort_values(ascending=False)
#
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='total_sales',data=df)
plt.title('Total Sales per Customer')
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='product',y='total_sales',data=df)
plt.title('Total Sales per Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='quantity',data=df)
plt.xlabel
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)
```

```python
data={
    'order':range(1,51),
    'customer':['customer'+str(i) for i in range(1,51)],
    'product':['product'+str(i) for i in range(1,51)],
    'quantity':np.random.randint(1,10,50),
    'price':np.random.randint(10,100,50),

}
df=pd.DataFrame(data)
df
```

```python
df['total_sales']=df['quantity']*df['price']
df
```

```python
df.isna().sum()
```

```python
df.loc[df['total_sales']>600]
```

```python
df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('customer')['quantity'].sum().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='total_sales',data=df)
plt.title('Total Sales per Customer')
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='product',y='total_sales',data=df)
plt.title('Total Sales per Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='quantity',data=df)
plt.title('Total Quantity per Customer')
plt.xlabel('Customer')
plt.ylabel('Total Quantity')
plt.show()
```
