```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.arange(11)
y=x**2
print(x,y)
plt.scatter(x,y)
plt.plot(x,y,'r-*')
plt.xlabel('x axis title')
plt.ylabel('y axis title')
plt.title('title')
plt.xticks(range(0,11))
plt.yticks(range(0,121,10))
plt.show()
```

```python
fig=plt.figure(figsize=(10,10))
axes=fig.add_axes([0,0,1,1])
axes.plot(x,y,'r-*')
axes.set_xlabel('x axis title')
axes.set_ylabel('y axis title')
axes.set_title('title')
axes.set_xticks(range(0,11))
axes.set_yticks(range(0,121,10))
plt.savefig('fig.png')
plt.show()
```

```python

fig,axes=plt.subplots(2,2,sharex= True,sharey=True,figsize=(8,5))
plt.tight_layout()
axes[0][0].plot(x,y,'r-*')
axes[0][0].set_xlabel('x axis title')
axes[0][0].set_ylabel('y axis title')
axes[0][0].set_title('title')
axes[0][0].set_xticks(range(0,11))
axes[0][0].set_yticks(range(0,121,10))
axes[0][1].plot(x,y,'r-*')
axes[0][1].set_xlabel('x axis title')
axes[0][1].set_ylabel('y axis title')
axes[0][1].set_title('title')
axes[0][1].set_xticks(range(0,11))
axes[0][1].set_yticks(range(0,121,10))
axes[1][1].scatter(x,y)
```

```python
fig3=plt.figure(figsize=(8,5))
axes3=fig3.add_axes([0,0,1,1])
axes_inset=axes3.inset_axes([0.2,0.5,0.4,0.4])
axes3.plot(x,y,'r-*')
axes_inset.plot(x,y,'r-*')
```

```python
fig3=plt.figure(figsize=(8,5))
axes_main=fig3.add_axes([0,0,1,1])
axes_inset=fig3.add_axes([.30,.05,0.3,0.4])
axes_main.plot(x,y,'r-*')
axes_inset.scatter(x,y)
```
