```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn
```

```python
data={
    'order':range(1,100001),
    'customer':[np.random.choice(['a','b','c','d',None]) for i in range(1,100001)],
    'product':[np.random.choice(['p1','p2','p3','p4',None]) for i in range(1,100001)],
    'quantity':[np.random.randint(1,10) for i in range(1,100001)],
    'price':[np.random.randint(10,100) for i in range(1,100001)],

}
df=pd.DataFrame(data)
print("raw data")
print(df)
```

```python
df['product']=df['product'].fillna('unknown')
df
```

```python
df['customer']=df['customer'].fillna('yes')
df
```

```python
df['total_price']=df['quantity']*df['price']
df
```
