```python
df['customer']=df['customer'].fillna('unknown')
df
```

```python
df['product']=df['product'].fillna('new')
df
```

```python
df['total_sales']=df['price']*df['quantity']
df
```

```python
top_revenue=df.sort_values(by='total_sales',ascending=False)
top_revenue
```

```python
top_product=df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
top_product
```

```python
top_customer=df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
top_customer
```

```python
df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
```

```python
least_revenue=df.sort_values(by='total_sales',ascending=True)
least_revenue
```

```python
least_sold=df.groupby('product')['quantity'].sum().sort_values(ascending=True)
least_sold
```

```python
best_product=df.groupby('product')['quantity'].sum().sort_values(ascending=False)
best_product
```

```python
wrost_product=df.groupby('product')['quantity'].sum().sort_values(ascending=True)
wrost_product
```
