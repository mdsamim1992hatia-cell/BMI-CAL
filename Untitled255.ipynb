```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
#
```

```python
data={
    'order':range(1,1001),
    'product':np.random.choice(['Apple','Ball','luchi',None,None],1000),
    'customer':np.random.choice(['Alex','Bin','Clli','Dall',None,None],1000),
    'quantity':np.random.randint(1,5,1000),
    'price':np.random.randint(10,50,1000),
}
print("====raw data=====")
df=pd.DataFrame(data)
df
```

```python
df['total_sales']=df['quantity']*df['price']
df
```
