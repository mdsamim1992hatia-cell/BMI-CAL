```python
tips=sns.load_dataset('tips')
tips
```

```python
tips.info()
```

```python
tips['total_bill'].hist(bins=30,color='red',edgecolor='black',)
```

```python
sns.histplot(tips['total_bill'],bins=30,color='red',edgecolor='black',kde=True)
```

```python
sns.histplot(tips['total_bill'],bins=30,color='red',edgecolor='black',kde=True,cumulative=True)
```

```python
sns.histplot(tips['total_bill'],bins=30,color='red',edgecolor='black',kde=False)
```
