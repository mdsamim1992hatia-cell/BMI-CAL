```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order_id':[101,102,103,104,105,106,107,108],
    'customer':['samim','adib','ajij','jamil','daed','jeaj','samim','samim',],
    'product':['laptop','dekstop','light','tv','werw','tv','laptop','tv'],
    'catagories':['accesories','glosary','electric','acess','samud','glosary','glosary','samud'],
    'quantity' :[1,2,3,6,5,6,8,9],
    'price':[100,555,444,555,5566,9996,23232,25252],
    'order_date':pd.date_range('20230101',periods=8)

}
df=pd.DataFrame(data)
df
```

```python
df['total_sales']=df['quantity']*df['price']
df
```

```python
total_revenue=df['total_sales'].sum()
total_revenue
```

```python
best_product=df.groupby('product')['total_sales'].sum().idxmax()
best_product
```

```python
categories_sales=df.groupby('catagories')['total_sales'].sum().sort_values(ascending=False)
categories_sales
```

```python
customer_spending=df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
customer_spending
```

```python
df['order_date']=pd.to_datetime(df['order_date'])
print(df.info())
```

```python
daily_sales=df.groupby(df['order_date'].dt.date)['total_sales'].sum()
daily_sales
```

```python
plt.figure(figsize=(10,6))
plt.plot(daily_sales.index,daily_sales.values,marker='o',linestyle='-')
plt.title('Daily Sales Trend')
plt.xlabel('Date')
plt.ylabel('Total Sales')
```

```python
daily_sales.plot(kind='bar',color='blue')
plt.title('Best Selling Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```
