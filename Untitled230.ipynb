```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'customer':range(1,11),
    'age':[25,30,35,40,None,50,None,60,65,None],
    'city':['New York','Los Angeles','Chicago','Houston','Phoenix','San Antonio','San Diego','Dallas','San Jose','Austin'],
    'gender':['M','F','M','F','M','F','M','F','M','F'],
    'purchase':[100,150,200,250,300,350,555,np.nan,np.nan,np.nan],
    'payment_method':['Credit Card','PayPal','Debit Card','Credit Card','PayPal','Debit Card','Credit Card','PayPal','Debit Card','Credit Card'],
}
df=pd.DataFrame(data)
print("===raw data===")
print(df)
print("===data info===")
print(df.info())
```

```python
df.isna().sum()
```

```python
df.isna().sum().sum()
```

```python
df['age'].mean()
```

```python
df['purchase']=df['purchase'].fillna(0)
df
```

```python
df['age']=df['age'].fillna(df['age'].mean())
df
```

```python

plt.figure(figsize=(10,6))
sns.barplot(x='city',y='purchase',data=df)
plt.title('purchase by city')
plt.xlabel('city')
plt.ylabel('purchase')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='gender',y='purchase',data=df)
plt.title('purchase by gender')
plt.xlabel('gender')
plt.ylabel('purchase')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='payment_method',y='purchase',data=df)
plt.title('purchase by payment_method')
plt.xlabel('payment_method')
plt.ylabel('purchase')
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='age',y='purchase',data=df)
plt.title('purchase by age')
plt.xlabel('age')
plt.ylabel('purchase')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='age',y='purchase',hue='gender',data=df)
plt.title('purchase by age and gender')
plt.xlabel('age')
plt.ylabel('purchase')
plt.show()
```
