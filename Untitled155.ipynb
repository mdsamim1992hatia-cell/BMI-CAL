```python
import pandas as pd
```

```python
data={
    'date':['2026-01-01','2026-01-02','2026-01-03','2026-01-04','2026-01-05'],
    'sales':[100,110,120,130,140],
    'prduct':['A','B','A','B','A'],
    'region':['North','South','East','West','North'],
    'quntity':[10,20,30,40,50]
}
df=pd.DataFrame(data)
df
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
plt.plot(df['date'],df['sales'],'r-*')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.title('Sales Over Time')
plt.grid(True)
```

```python
fig=plt.figure(figsize=(10,6))
for product,group in df.groupby('prduct'):
    plt.plot(group['date'],group['sales'],marker='o',label=product)
    plt.xlabel('Date')
    plt.ylabel('Sales')
    plt.title('Sales Over Time by Product')
    plt.legend()
    plt.grid(True)

```
