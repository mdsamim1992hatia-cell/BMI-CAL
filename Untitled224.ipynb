```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':range(1,11),
    'customer':['A','B','C',None,'E','F','G','H',None,'J'],
    'product':['X','Y',None,'X','Y',None,'X','Y',None,'X'],
    'quantity':[None,3,2,4,None,2,3,1,None,5],
    'price':[10,15,20,12,None,16,15,None,55,55]
}
df=pd.DataFrame(data)
print("Dataframe:")
print(df)
```

```python
df.isna().sum()
```

```python
df.isna().sum().sum()
```

```python
df['customer'].fillna('Unknown',inplace=True)
df
```

```python
df['product'].fillna('Unknown',inplace=True)
df
```

```python
df['price']=df['price'].fillna(df['price'].mean())
df
```

```python
df['quantity']=df['quantity'].fillna(df['quantity'].mean())
df
```

```python
df.isna().sum()
```

```python
df[['product','customer']]
```

```python
df['total_sales']=df['quantity']*df['price']
df
```

```python
df.groupby('customer')['total_sales'].sum().sort_values(ascending=False)
df.groupby('product')['total_sales'].sum().sort_values(ascending=False)
df.groupby(['customer','product'])['total_sales'].sum()
print(df.groupby(['customer','product'])['total_sales'].sum())
df.groupby(['customer','product'])['total_sales'].sum().unstack()
df.groupby(['customer','product'])['total_sales'].sum().unstack().fillna(0)

```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='total_sales',data=df)
plt.title('Total Sales by Customer')
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='product',y='total_sales',data=df)
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```
