```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data={
    'order':['a','b','c','d','e'],
    'customer':['alex','drank','alex','koil','alex'],
    'product':['apple','orange',None,None,'orange'],
    'quantity':[1,2,3,4,5],
    'price':[10,20,30,40,50]
}
df=pd.DataFrame(data)
print("===RAW DATA===")
print(df)
```

```python
df['product']=df['product'].fillna('unknown')
print("===CLEAN DATA===")
print(df)
```

```python
df['totll_sales']=df['quantity']*df['price']
print("===TOTALL SALES===")
print(df)
```

```python
df.groupby('customer')['totll_sales'].sum().sort_values(ascending=False)


```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='totll_sales',data=df)
plt.title('Total Sales by Customer')
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.show()
```

```python
df.groupby('product')['totll_sales'].sum().sort_values(ascending=False)
```

```python
plt.figure()
sns.barplot(x='product',y='totll_sales',data=df)
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()


```

```python
df.groupby('order')['totll_sales'].sum().sort_values(ascending=False)
```

```python
plt.figure()
sns.barplot(x='order',y='totll_sales',data=df)
plt.title('Total Sales by Order')
plt.xlabel('Order')
plt.ylabel('Total Sales')
plt.show()
```

```python
df.groupby('quantity')['totll_sales'].sum().sort_values(ascending=False)
```

```python
plt.figure()
sns.barplot(x='quantity',y='totll_sales',data=df)
plt.title('Total Sales by Quantity')
plt.xlabel('Quantity')
plt.ylabel('Total Sales')
plt.show()
```

```python
data1={
    'order':['a','b','c','d','e','a','b','c','d','e','a','b','c','d','e'],
    'customer':['alex','drank','alex',None,'alex','alex','drank','alex','koil','alex','alex','drank','alex','koil','alex'],
    'product':['apple','orange',None,None,'orange','apple','orange',None,None,'orange','apple','orange',None,None,'lyci'],

    'quantity':[1,2,3,4,5,None,2,3,4,None,1,2,3,4,5],
    'price':[10,20,30,40,50,10,20,None,40,50,10,None,30,40,50]

}
df=pd.DataFrame(data1)
print("===RAW DATA===")
print(df)
```

```python
df['customer']=df['customer'].fillna('unknown')
df['product']=df['product'].fillna('unknown')
df['quantity']=df['quantity'].fillna(0)
df['price']=df['price'].fillna(0)
print("===CLEAN DATA===")
print(df)
```

```python
df['totall_sales']=df['quantity']*df['price']
print("===TOTALL SALES===")
print(df)
```

```python
top_customers=df.groupby('customer')['totall_sales'].sum().sort_values(ascending=False)
print("===TOP CUSTOMERS===")
print(top_customers)
top_products=df.groupby('product')['totall_sales'].sum().sort_values(ascending=False)
print("===TOP PRODUCTS===")
print(top_products)
```

```python
median_sales=df['totall_sales'].median()
high_sales_orders=df[df['totall_sales']>median_sales]
print("===HIGH SALES ORDERS===")
print(high_sales_orders)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x='customer',y='totall_sales',data=df)
plt.title('Total Sales by Customer')
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='product', y='totall_sales', data=df)
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x=high_sales_orders['order'], y=high_sales_orders['totall_sales'], data=df)
plt.title('High Sales Orders')
plt.xlabel('Order')
plt.ylabel('Total Sales')
plt.show()
```
