```python
df1=pd.DataFrame({
    'key':['a','b','c','d'],
    'a1':[1,2,3,4],
    'b1':[5,6,7,8]
})
df1
```

```python
df2=pd.DataFrame({
    'key':['a','b','c','d'],
    'c1':[9,555,33,44],
    'd1':[55,66,77,88]
})
df2
```

```python
pd.merge(df1,df2,on='key',how='inner')
```

```python
pd.merge(df1,df2,on='key',how='outer')[['key','a1','c1']]
```

```python
pd.merge(df1,df2,on='key',how='left')
```

```python
pd.merge(df1,df2,on='key',how='right')
```

```python
pd.merge(df1,df2,on='key',how='outer',indicator=True)
```
