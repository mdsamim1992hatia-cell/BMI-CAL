```python
data={
    'order':range(1,11),
    'store':['dhaka','khulna','rajsahi','dhaka','khulna','noakhali','rongpur','akali','dhaka','khulna'],
    'customer':['alex','noamm','reag','alex','noamm','reag','alex','noamm','reag','alex'],
    'quantity':[55,55,5252,2525,2525,232,np.nan,np.nan,np.nan,np.nan],
    'price':[222,55,55,55,5252,2525,252,None,None,None],
    'data':pd.date_range('20220101',periods=10)

}
df=pd.DataFrame(data)
print("===raw data===")
print(df)
```

```python
df.isna().sum()
```

```python
df['price']=df['price'].fillna(0)
df
```

```python
df['quantity']=df['quantity'].fillna(df['quantity'].mean())
df
```

```python
df['data']=pd.to_datetime(df['data'])
df
```

```python
store_total_sales=df.groupby('store')['quantity'].sum()
print(store_total_sales)
```

```python
df['total_sales']=df['quantity']*df['price']
print(df)
```

```python
store_avg_sales=df.groupby('store')['total_sales'].mean()
print(store_avg_sales)
```

```python
daily_sales=df.groupby('data')['total_sales'].sum()
print(daily_sales)
```

```python
top_3_stores=store_total_sales.nlargest(3)
print(top_3_stores)
```

```python
top_3_order=df.nlargest(3,'total_sales')
print(top_3_order)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=store_total_sales.index,y=store_total_sales.values)
plt.xlabel('Store')
plt.ylabel('Total Sales')
plt.title('Total Sales by Store')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.lineplot(x=daily_sales.index,y=daily_sales.values)
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.title('Daily Sales')
```
