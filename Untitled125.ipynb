```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.DataFrame({
    'store':['a','b','c','a','b','c'],
    'sales':[100,2225,2222,555,555,555]
})
df.groupby('store')['sales'].agg(['sum','mean','max','min'])
```

```python
df.groupby('store')['sales'].mean()
```

```python
df.groupby('store')['sales'].sum().sort_values(ascending=False)
```

```python
df=pd.DataFrame({
    'dept':['it','hr','ceo'],
    'salary':[1000,2000,3000]
})
df.groupby('dept')['salary'].describe()
```
